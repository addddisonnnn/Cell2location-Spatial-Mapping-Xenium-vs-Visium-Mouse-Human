```python
# Initial sanity checks after downloading the human GBM visium dataset
import scanpy as sc
import numpy as np

gbm_dir = "data/humanGlioblastoma/"

# Load and immediately fix variable names
adata_gbm_vis = sc.read_visium(
    path=gbm_dir,
    count_file="Parent_Visium_Human_Glioblastoma_filtered_feature_bc_matrix.h5"
)
adata_gbm_vis.var_names_make_unique()

# Confirm
print("GBM Visium shape:", adata_gbm_vis.shape)
print("Are counts raw integers:", 
      np.all(adata_gbm_vis.X.data == adata_gbm_vis.X.data.astype(int)))
print("Spatial image available:", "spatial" in adata_gbm_vis.uns)
print("Number of spots in tissue:", adata_gbm_vis.obs["in_tissue"].sum())
```

```python
# Initial sanity checks after downloading the Xenium dataset
import scanpy as sc
import pandas as pd
import numpy as np
import json

xenium_dir = "data/xeniumMouseBrain/"

# Load cell by gene matrix
adata_xenium = sc.read_10x_h5(xenium_dir + "cell_feature_matrix.h5")
adata_xenium.var_names_make_unique()
print("Xenium data shape (cells x genes):", adata_xenium.shape)

# Load cell coordinates
cells = pd.read_csv(xenium_dir + "cells.csv.gz")
print("\nCells CSV columns:", cells.columns.tolist())
print(cells.head())

# Attach coordinates
adata_xenium.obs["x_centroid"] = cells["x_centroid"].values
adata_xenium.obs["y_centroid"] = cells["y_centroid"].values

# Check counts are raw integers
print("\nAre counts raw integers:", 
      np.all(adata_xenium.X.data == adata_xenium.X.data.astype(int)))

# Check gene panel
with open(xenium_dir + "gene_panel.json") as f:
    panel = json.load(f)
print("\nNumber of genes in panel:", len(panel["payload"]["targets"]))

# Quick summary
print("\n--- Xenium Summary ---")
print(f"Cells: {adata_xenium.n_obs}")
print(f"Genes: {adata_xenium.n_vars}")
print(f"Mean transcripts per cell: {adata_xenium.X.sum(axis=1).mean():.1f}")
```
